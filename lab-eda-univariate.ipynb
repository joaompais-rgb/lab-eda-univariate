## Parte 1. Categorías de producto

**Pregunta de negocio:** ¿qué categorías tienen más productos en Amazon UK y cómo se comparan?

> **Pergunta de negócio:** que categorias têm mais produtos na Amazon UK e como se comparam?

**Cómo funciona**

1. `value_counts()` da la tabla de frecuencias: cuántos productos hay en cada categoría, de más a menos.
2. `value_counts(normalize=True) * 100` da el mismo reparto en %.
3. Se juntan las dos en una tabla con `pd.DataFrame` y se enseñan las 5 primeras con `head(5)`.
4. `nunique()` cuenta cuántas categorías distintas hay.

**Por qué así:** la tabla de frecuencias es la forma de describir una variable categórica (lección `1.EDA_univariate`); el % permite comparar categorías aunque el total sea enorme.

**Herramientas:** `value_counts`, `normalize`, `pd.DataFrame`, `head`, `nunique`.

> **Como funciona (português)**
>
> 1. O `value_counts()` dá a tabela de frequências: quantos produtos há em cada categoria, do maior para o menor.
> 2. O `value_counts(normalize=True) * 100` dá a mesma repartição em %.
> 3. Juntam-se as duas numa tabela com `pd.DataFrame` e mostram-se as 5 primeiras com `head(5)`.
> 4. O `nunique()` conta quantas categorias diferentes há.
>
> **Porquê assim:** a tabela de frequências é a forma de descrever uma variável categórica; a % permite comparar categorias mesmo com um total enorme.

```python
frecuencias = df["category"].value_counts()
proporciones = df["category"].value_counts(normalize=True) * 100
tabla_categorias = pd.DataFrame({"productos": frecuencias, "pct": proporciones.round(2)})

print("Categorías distintas:", df["category"].nunique())
tabla_categorias.head(5)
```

**➜ Qué nos llevamos:** las 5 categorías con más productos y su peso en el catálogo.

> **➜ O que tiramos:** as 5 categorias com mais produtos e o seu peso no catálogo.

3. Para el gráfico de tarta se usan las 5 primeras categorías y se juntan todas las demás en «Otras», para ver qué parte del total ocupan.
4. `plot.pie(autopct="%1.1f%%")` escribe el % en cada porción, como en la lección.

## Parte 2. Precios

**Pregunta de negocio:** ¿cómo se reparten los precios y hay rangos de precio más habituales?

> **Pergunta de negócio:** como se distribuem os preços e há gamas de preço mais habituais?

**Cómo funciona**

1. **Centralidad:** media (`mean`), mediana (`median`) y moda (`mode()[0]`, el precio más repetido), como en la lección.
2. **Dispersión:** varianza (`var`), desviación típica (`std`), rango (máximo − mínimo) y rango intercuartílico (IQR = cuartil 3 − cuartil 1, con `quantile`).
3. Se guardan en un diccionario y se enseñan como tabla con `pd.Series`.

**Por qué así:** la media sola engaña cuando hay precios muy altos; compararla con la mediana y la moda dice si la distribución está sesgada. El IQR mide la dispersión del 50 % central sin que le afecten los extremos.

**Herramientas:** `mean`, `median`, `mode`, `var`, `std`, `max`, `min`, `quantile`, `pd.Series`, `round`.

> **Como funciona (português)**
>
> 1. **Centralidade:** média (`mean`), mediana (`median`) e moda (`mode()[0]`, o preço mais repetido), como na lição.
> 2. **Dispersão:** variância (`var`), desvio padrão (`std`), amplitude (máximo − mínimo) e amplitude interquartil (IQR = quartil 3 − quartil 1, com `quantile`).
> 3. Guardam-se num dicionário e mostram-se como tabela com `pd.Series`.
>
> **Porquê assim:** a média sozinha engana quando há preços muito altos; compará-la com a mediana e a moda diz se a distribuição está enviesada. O IQR mede a dispersão dos 50 % centrais sem ser afectado pelos extremos.

```python
precio = df["price"]
q1_precio = precio.quantile(0.25)
q3_precio = precio.quantile(0.75)

resumen_precio = pd.Series({
    "media": precio.mean(),
    "mediana": precio.median(),
    "moda": precio.mode()[0],
    "varianza": precio.var(),
    "desviacion_tipica": precio.std(),
    "minimo": precio.min(),
    "maximo": precio.max(),
    "rango": precio.max() - precio.min(),
    "q1": q1_precio,
    "q3": q3_precio,
    "iqr": q3_precio - q1_precio,
})
resumen_precio.round(2)
```

**➜ Qué nos llevamos:** el precio típico (mediana y moda) frente a la media, y cuánto varían los precios.

> **➜ O que tiramos:** o preço típico (mediana e moda) contra a média, e quanto variam os preços.

**Cómo funciona**

1. Histograma de todos los precios con `sns.histplot`. Con unos pocos productos muy caros, casi todo queda en la primera barra y no se lee: es lo que avisa el enunciado.
2. **Solución:** repetir el histograma solo con los precios hasta el percentil 99 (`quantile(0.99)`), que deja fuera el 1 % más caro y enseña dónde están la mayoría.
3. Diagrama de caja (`sns.boxplot`) de todos los precios para ver los valores atípicos.

**Por qué así:** el histograma enseña la forma y la caja los atípicos (lección `1.EDA_univariate`). Recortar al percentil 99 solo cambia el gráfico, no los datos.

**Herramientas:** `sns.histplot`, `quantile`, filtros, `sns.boxplot`, `plt.subplots`.

> **Como funciona (português)**
>
> 1. Histograma de todos os preços com `sns.histplot`. Com uns poucos produtos muito caros, quase tudo fica na primeira barra e não se lê: é o que avisa o enunciado.
> 2. **Solução:** repetir o histograma só com os preços até ao percentil 99 (`quantile(0.99)`), que deixa de fora o 1 % mais caro e mostra onde está a maioria.
> 3. Caixa de bigodes (`sns.boxplot`) de todos os preços para ver os valores atípicos.
>
> **Porquê assim:** o histograma mostra a forma e a caixa os atípicos. Cortar no percentil 99 só muda o gráfico, não os dados.

```python
p99_precio = precio.quantile(0.99)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 4))
sns.histplot(precio, bins=50, ax=ax1)
ax1.set_title("Todos los precios")
sns.histplot(precio[precio <= p99_precio], bins=50, ax=ax2)
ax2.set_title(f"Precios hasta el percentil 99 ({p99_precio:.0f} £)")
plt.show()

plt.figure(figsize=(10, 3))
sns.boxplot(x=precio, color="lightblue")
plt.title("Precios: diagrama de caja")
plt.show()
```

**➜ Qué nos llevamos:** en qué rango está la mayoría de los productos y cuántos precios son atípicos: el informe final lo resume.

> **➜ O que tiramos:** em que gama está a maioria dos produtos e quantos preços são atípicos: o relatório final resume-o.

## Parte 3. Valoraciones (`stars`)

**Pregunta de negocio:** ¿cómo puntúan los clientes y hay alguna tendencia?

> **Pergunta de negócio:** como pontuam os clientes e há alguma tendência?

**Cómo funciona**

1. Las mismas medidas que con el precio: media, mediana y moda; varianza, desviación típica e IQR.
2. **Forma:** asimetría (`skew`: positiva = cola a la derecha, negativa = cola a la izquierda) y curtosis (`kurtosis`: mayor que 0 = más valores en los extremos que una normal).
3. Cuenta los productos con 0 estrellas: en este dataset, 0 suele significar «sin opiniones», no «muy mal valorado». Se repiten la media y la mediana sin ellos.

**Por qué así:** la asimetría y la curtosis dicen si la valoración se parece a una normal (lección `1.EDA_univariate`). Separar los 0 evita sacar conclusiones de productos que nadie ha valorado.

**Herramientas:** `mean`, `median`, `mode`, `var`, `std`, `quantile`, `skew`, `kurtosis`, filtros.

> **Como funciona (português)**
>
> 1. As mesmas medidas que no preço: média, mediana e moda; variância, desvio padrão e IQR.
> 2. **Forma:** assimetria (`skew`: positiva = cauda à direita, negativa = cauda à esquerda) e curtose (`kurtosis`: maior que 0 = mais valores nos extremos do que uma normal).
> 3. Conta os produtos com 0 estrelas: neste dataset, 0 costuma significar «sem opiniões», não «muito mal avaliado». Repetem-se a média e a mediana sem eles.
>
> **Porquê assim:** a assimetria e a curtose dizem se a valorização se parece com uma normal. Separar os 0 evita tirar conclusões de produtos que ninguém avaliou.

```python
estrellas = df["stars"]
resumen_estrellas = pd.Series({
    "media": estrellas.mean(),
    "mediana": estrellas.median(),
    "moda": estrellas.mode()[0],
    "varianza": estrellas.var(),
    "desviacion_tipica": estrellas.std(),
    "iqr": estrellas.quantile(0.75) - estrellas.quantile(0.25),
    "asimetria": estrellas.skew(),
    "curtosis": estrellas.kurtosis(),
})
display(resumen_estrellas.round(2))

sin_valorar = (estrellas == 0).mean() * 100
valorados = estrellas[estrellas > 0]
print(f"Productos con 0 estrellas: {sin_valorar:.1f} %")
print(f"Solo productos valorados: media {valorados.mean():.2f}, mediana {valorados.median():.2f}")
```

**➜ Qué nos llevamos:** cómo valoran los clientes y cuánto pesan los productos sin valorar en esas medidas.

> **➜ O que tiramos:** como avaliam os clientes e quanto pesam os produtos sem avaliação nessas medidas.

**Cómo funciona**

1. Histograma de las estrellas (todas) y, al lado, solo de los productos valorados (más de 0).

**Por qué así:** con muchos productos en 0, el primer histograma esconde cómo se reparten las valoraciones reales; el segundo lo enseña.

**Herramientas:** `plt.subplots`, `sns.histplot`.

> **Como funciona (português)**
>
> 1. Histograma das estrelas (todas) e, ao lado, só dos produtos avaliados (mais de 0).
>
> **Porquê assim:** com muitos produtos em 0, o primeiro histograma esconde como se repartem as avaliações reais; o segundo mostra-o.

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 4))
sns.histplot(estrellas, bins=50, ax=ax1)
ax1.set_title("Estrellas: todos los productos")
sns.histplot(valorados, bins=40, ax=ax2)
ax2.set_title("Estrellas: solo productos valorados")
plt.show()
```

**➜ Qué nos llevamos:** cuál es la valoración más habitual entre los productos que sí tienen opiniones.

> **➜ O que tiramos:** qual é a avaliação mais habitual entre os produtos que têm opiniões.

## 4. Informe para negocio

La celda siguiente escribe el informe con los resultados de las celdas anteriores: f-strings para los números y condiciones `if` para las frases que dependen de ellos (por ejemplo, si la media de precio es mucho mayor que la mediana, la distribución está sesgada). Así el informe siempre cuadra con los datos.

> **Relatório para o negócio.** A célula seguinte escreve o relatório com os resultados das células anteriores (f-strings e condições `if`), por isso bate sempre certo com os dados.

**Cómo funciona**

1. Toma las cifras ya calculadas (tablas y resúmenes) y las escribe en frases con f-strings.
2. Usa `if` para elegir la frase: por ejemplo, si la primera categoría tiene más del 25 % de los productos, dice que domina; si no, que el catálogo está repartido.
3. Imprime el informe por partes.

**Por qué así:** un informe escrito a mano puede quedar desfasado si cambian los datos; generado desde los resultados, no.

**Herramientas:** f-strings, `if ... else`, `print`.

> **Como funciona (português)**
>
> 1. Pega nos números já calculados e escreve-os em frases com f-strings.
> 2. Usa `if` para escolher a frase: por exemplo, se a primeira categoria tiver mais de 25 % dos produtos, diz que domina; se não, que o catálogo está repartido.
> 3. Imprime o relatório por partes.
>
> **Porquê assim:** um relatório escrito à mão pode ficar desactualizado se os dados mudarem; gerado a partir dos resultados, não.

```python
primera = tabla_categorias.index[0]
pct_primera = tabla_categorias["pct"].iloc[0]
pct_top_5 = tabla_categorias["pct"].head(5).sum()

print("1. CATEGORÍAS")
print(f"   - Hay {df['category'].nunique()} categorías. La que más productos tiene es «{primera}» ({pct_primera:.1f} %).")
print(f"   - Las 5 primeras suman el {pct_top_5:.1f} % del catálogo: {', '.join(tabla_categorias.index[:5])}.")
if pct_primera > 25:
    print("   - Una categoría domina el catálogo: conviene analizarla por separado.")
else:
    print("   - Ninguna categoría domina: el catálogo está muy repartido.")

print("\n2. PRECIOS")
print(f"   - Precio medio {resumen_precio['media']:.2f} £, mediano {resumen_precio['mediana']:.2f} £ y más repetido (moda) {resumen_precio['moda']:.2f} £.")
if resumen_precio["media"] > 1.5 * resumen_precio["mediana"]:
    print("   - La media es mucho mayor que la mediana: unos pocos productos muy caros tiran de ella. El precio típico es la mediana.")
print(f"   - La mitad central de los productos cuesta entre {resumen_precio['q1']:.2f} y {resumen_precio['q3']:.2f} £ (IQR {resumen_precio['iqr']:.2f} £).")
print(f"   - El 99 % de los productos cuesta {p99_precio:.0f} £ o menos; el máximo llega a {resumen_precio['maximo']:,.0f} £.")

print("\n3. VALORACIONES")
print(f"   - El {sin_valorar:.1f} % de los productos tiene 0 estrellas (sin opiniones).")
print(f"   - Entre los valorados, la media es {valorados.mean():.2f} y la mediana {valorados.median():.2f}: los clientes que opinan puntúan alto.")
if resumen_estrellas["asimetria"] > 0:
    lado = "a la derecha"
else:
    lado = "a la izquierda"
print(f"   - Asimetría {resumen_estrellas['asimetria']:.2f} (cola {lado}) y curtosis {resumen_estrellas['curtosis']:.2f}.")
print("   - La valoración no es normal: hay dos grupos, los productos sin opiniones (0) y los valorados, casi todos por encima de 3.")

print("\n4. RECOMENDACIONES")
print("   - Para fijar precios, comparar con la mediana de la categoría, no con la media.")
print("   - Conseguir las primeras opiniones es clave: muchos productos no tienen ninguna.")
```

**➜ Qué nos llevamos:** el informe de negocio que pide el enunciado, siempre coherente con los datos.

> **➜ O que tiramos:** o relatório de negócio que o enunciado pede, sempre coerente com os dados.
